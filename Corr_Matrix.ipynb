```python
import pandas as pd

data = pd.read_csv("C:/Users/amiru/hs4002_ay26/Christian_Universities_as_Moral_Communities_Drinking,_Sex,_and_Drug.csv")

data.corr().round(3)

data




```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Compute correlation matrix
Corr_Matrix = data.corr().round(2)

# Plot heatmap
plt.figure(figsize=(100, 60))
sns.heatmap(Corr_Matrix, annot=True, cmap="coolwarm", center=0)

# Save directly as PDF in your project folder
plt.savefig("C:/Users/amiru/hs4002_ay26/heatmap.pdf", format="pdf", dpi=400)

plt.show()


```
